```python

import os
import kagglehub
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, OneHotEncoder, LabelEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.svm import SVC
from sklearn.tree import DecisionTreeClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    log_loss,
    confusion_matrix,
    ConfusionMatrixDisplay,
)

# 1. Direct import from Kaggle using kagglehub
path = kagglehub.dataset_download("rabieelkharoua/predict-online-gaming-behavior-dataset")
csv_file = os.path.join(path, "online_gaming_behavior_dataset.csv")
df = pd.read_csv(csv_file)

# Drop PlayerID as it is a unique identifier, not a predictive feature
if "PlayerID" in df.columns:
    df = df.drop(columns=["PlayerID"])

# Stratified sample of 8,000 rows so SVM with probability=True runs in seconds on Colab.
# Set MAX_SAMPLES = None to run on all 40,034 rows.
MAX_SAMPLES = 8000
if MAX_SAMPLES is not None and len(df) > MAX_SAMPLES:
    df, _ = train_test_split(
        df, train_size=MAX_SAMPLES, stratify=df["EngagementLevel"], random_state=42
    )

X = df.drop(columns=["EngagementLevel"])
y_raw = df["EngagementLevel"]

# Encode target labels ('High', 'Low', 'Medium') into integers
le = LabelEncoder()
y = le.fit_transform(y_raw)
class_names = le.classes_

print("Dataset: Predict Online Gaming Behavior")
print(f"Total samples  : {X.shape[0]}")
print(f"Total features : {X.shape[1]}")
print(f"Target classes : {list(class_names)}")

# 2. Identify numerical and categorical columns
categorical_cols = ["Gender", "Location", "GameGenre", "GameDifficulty"]
numerical_cols = [col for col in X.columns if col not in categorical_cols]

# 3. Split into 80% training and 20% testing sets (stratified)
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42, stratify=y
)
print(f"Training samples: {len(X_train)}")
print(f"Testing samples : {len(X_test)}")

# 4. Preprocessing pipelines
scaled_preprocessor = ColumnTransformer(
    transformers=[
        ("num", StandardScaler(), numerical_cols),
        ("cat", OneHotEncoder(drop="first", sparse_output=False), categorical_cols),
    ]
)

unscaled_preprocessor = ColumnTransformer(
    transformers=[
        ("num", "passthrough", numerical_cols),
        ("cat", OneHotEncoder(drop="first", sparse_output=False), categorical_cols),
    ]
)

# 5. Initialize the four classifiers
classifiers = {
    "SVM": Pipeline([
        ("prep", scaled_preprocessor),
        ("clf", SVC(kernel="rbf", probability=True, random_state=42)),
    ]),
    "Decision Tree": Pipeline([
        ("prep", unscaled_preprocessor),
        ("clf", DecisionTreeClassifier(max_depth=8, random_state=42)),
    ]),
    "KNN": Pipeline([
        ("prep", scaled_preprocessor),
        ("clf", KNeighborsClassifier(n_neighbors=7)),
    ]),
    "Naive Bayes": Pipeline([
        ("prep", scaled_preprocessor),
        ("clf", GaussianNB()),
    ]),
}

# 6. Train models and compute evaluation metrics
metrics_list = []
preds_dict = {}

for model_name, clf in classifiers.items():
    print(f"Training {model_name}...")
    clf.fit(X_train, y_train)
    y_pred = clf.predict(X_test)
    y_prob = clf.predict_proba(X_test)

    metrics_list.append({
        "Model": model_name,
        "Accuracy": accuracy_score(y_test, y_pred),
        "Precision": precision_score(y_test, y_pred, average="macro"),
        "Recall": recall_score(y_test, y_pred, average="macro"),
        "F1-Score": f1_score(y_test, y_pred, average="macro"),
        "Log Loss": log_loss(y_test, y_prob),
    })
    preds_dict[model_name] = y_pred

# 7. Display and export the comparison table
results_df = pd.DataFrame(metrics_list)
print("\n=== MODEL PERFORMANCE COMPARISON ===")
print(results_df.round(4).to_string(index=False))
results_df.to_csv("gaming_model_comparison.csv", index=False)

# 8. Plot 1: Accuracy, Precision, Recall, and F1-Score comparison
results_df.set_index("Model")[
    ["Accuracy", "Precision", "Recall", "F1-Score"]
].plot(kind="bar", figsize=(10, 6), edgecolor="black")
plt.title("Classification Models Comparison (Online Gaming Behavior Dataset)")
plt.ylabel("Score")
plt.ylim(0, 1.08)
plt.xticks(rotation=0)
plt.legend(loc="lower right")
plt.grid(axis="y", linestyle="--", alpha=0.5)
plt.tight_layout()
plt.savefig("metrics_comparison.png", dpi=300)
plt.show()

# 9. Plot 2: Log Loss comparison (lower is better)
plt.figure(figsize=(8, 5))
bars = plt.bar(
    results_df["Model"],
    results_df["Log Loss"],
    color=["#2b5c8f", "#d95f02", "#1b9e77", "#7570b3"],
    edgecolor="black",
)
plt.title("Log Loss Comparison Across Models (Lower is Better)")
plt.ylabel("Log Loss")
plt.grid(axis="y", linestyle="--", alpha=0.5)
for bar in bars:
    height = bar.get_height()
    plt.text(
        bar.get_x() + bar.get_width() / 2,
        height + 0.02,
        f"{height:.4f}",
        ha="center",
        va="bottom",
    )
plt.tight_layout()
plt.savefig("log_loss_comparison.png", dpi=300)
plt.show()

# 10. Plot 3: All four Confusion Matrices in a single 2x2 grid
fig, axes = plt.subplots(2, 2, figsize=(10, 8))
for ax, (model_name, y_pred) in zip(axes.ravel(), preds_dict.items()):
    cm = confusion_matrix(y_test, y_pred)
    disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=class_names)
    disp.plot(ax=ax, cmap="Blues", colorbar=False)
    ax.set_title(f"Confusion Matrix - {model_name}")
plt.tight_layout()
plt.savefig("confusion_matrices.png", dpi=300)
plt.show()

# 11. Print best model per metric
print("\nBest Accuracy  :", results_df.loc[results_df["Accuracy"].idxmax(), "Model"])
print("Best F1-Score  :", results_df.loc[results_df["F1-Score"].idxmax(), "Model"])
print("Lowest Log Loss:", results_df.loc[results_df["Log Loss"].idxmin(), "Model"])
```
